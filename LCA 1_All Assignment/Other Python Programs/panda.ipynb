```python
import pandas as pd #how to use pandas
marks=pd.Series([99,98,97,96,95])
print(marks)
print(marks[2])
marks1=pd.Series(
    [20,30,40],
    index=["A","B","C"]) #here we are changing the index to A,B,C form instead of 0,1,2.....
print(marks1["B"])
```

```python
#dataframe is a table with multiple rows
import pandas as pd
data={'name':["A","B","C"],
"age":[30,20,15],
"marks":[20,30,40]}
dataframe=pd.DataFrame(data) #curly brackets because in dataframe there are multiple sets so we write in dictionary format
print(dataframe)  
print(dataframe["age"])
```
