```python
#List_Problem 1
items=["Milk","Bread","Eggs","Butter"]
#to add cheese at the end
items.append("Cheese")
#to insert juice at position 2
items.insert(1,"Juice")
#to change the entry Eggs to chocolate
items[3]="Chocolate"
#to delete the entry Bread
items.remove("Bread")
#to display the first 3 items using slicing
print(items[0:3])
#to display the entire list
print(items)

#Problem 2_ Student Marks
marks=[78,85,92,67,88]
#to add a new mark 95
marks.append(95)
#insert 80 at second position
marks.insert(1,80)
#to change 67 to 72
marks[4]=72
#to remove 85
marks.remove(85)
#to display the first 3 marks
print(marks[0:3])
#to find the maximum and minimum marks
print(max(marks))
print(min(marks))


#Problem3_Cricket Team
players=["Rohit","Virat","Rahul","Hardik","Bumrah"]
#to add Surya to the team
players.append("Surya")
#to insert Dhoni at the beginning
players.insert(0,"Dhoni")
#to replace Rahul with Pant
players[3]="Pant"
#to remove Hardik
players.remove("Hardik")
#to display the first three player names
print(players[0:3])
#to sort the names alphabetically
print(sorted(players))


#Problem4_To-Do List
tasks=["Study","Excercise","Complete Assignment"]
#to append "Read Book"
tasks.append("Read Book")
#to insert "Breakfast" at the beginning
tasks.insert(0,"Breakfast")
#to change "Excercise" to "Walking"
tasks[2]="Walking"
#to delete Read Book
tasks.remove("Read Book")
#to display the last two tasks using slicing
print(tasks[-2:])
#to print the number of tasks
print(len(tasks))

#Problem5_Student Information
student=("Amit",18,"Computer Engineering",85)
#to display the student's name
print(student[0])
#to display the age
print(student[1])
#to display the last two elements using slicing
print(student[-2:])
#to find the length of the tuple
print(len(student))
#to check whether "Computer Engineering" is present
if "Computer Engineering" in student:
    print("Computer Engineering is present")
else:
    print("Computer Engineering is not present")






```
