```python
#Problem6_Days of the week
days=("Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday")
#to display the first 3 days
print(days[:3])
#to display the last two days
print(days[-2:])
#to display Wednesday
print(days[2])
#to find the number of days
print(len(days))
#to check whether "Sunday" exists
if "Sunday" in days:
    print("Sunday exists in the tuple")
else:
    print("Sunday doesz not exist in the tuple")
#to find the position of friday
print(days.index("Friday"))

#Problem7_ Marks Tuple
marks=(78,90,65,88,90,72)
#to find the highest mark
print(max(marks))
#to find the lowest mark
print(min(marks))
#to count how many times 90 occurs
print(marks.count(90))
#to find the position of 88
print(marks.index(88))
#to display the first four marks using slicing
print(marks[:4])

#Problem8_ Student Details
student={"name":"Neha","age":18,"course":"Computer Engineering"}
#to add "city":"Pune"
student["city"]="Pune"
#Update the age to 19
student["age"]=19
#to change the course to "AI Engineering"
student["course"]="AI Engineering"
#to delete city
del student["city"]
#to display all keys
print(student.keys())
#to display all values
print(student.values())
#to display the complete dictionary
print(student)

#Problem9_Mobile Phone
mobile={"brand":"Samsung","model":"S24","price":70000}
#to add the colour black
mobile["colour"]="Black"
#to update the price to 65000
mobile["price"]=65000
#to change the model to "S25"
mobile["model"]="S25"
#to delete colour
del mobile["colour"]
#to display all keys and values
print(mobile.items())
#to check whether "price" exists
if "price" in mobile:
    print("Price exists")
else:
    print("Price does not exist")

#Problem10_Employee Record
employee={"name":"Raj","department":"IT","salary":50000}
#to add "experience":3
employee["experience"]=3
#to update salary to 55000
employee["salary"]=55000
#to change department to Cyber Security
employee["department"]="Cyber Security"
#to delete experience
del employee["experience"]
#to print the employee's name and salary
print(employee["name"],employee["salary"])
#to print all dictionary items
print(employee.items())

#Problem11_College Canteen
food=["Pizza","Burger","Sandwich","Pasta"]
food.append("Fries")
#to insert noodles at position 2
food.insert(1,"Noodles")
#to remove Burger
food.remove("Burger")
#to replace Pasta with Biryani
food[3]="Biryani"
#to display items from position 2 to 4
print(food[1:4])
#to sort the list
print(sorted(food))









```
